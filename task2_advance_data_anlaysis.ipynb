### Merge-left join,right join,inner join,outer join

#### 1.inner_join

```python
sales_edited_data=pd.merge(sales_data_2017,sales_data_2018,on=["Party","Product"],how="inner",sort=False)
sales_edited_data
```

#### 2.outer_join

```python
sales_edited_data=pd.merge(sales_data_2017,sales_data_2018,on=["Party","Product"],how="outer",sort=False)
sales_edited_data
```

#### 3.left_join

```python
sales_edited_data=pd.merge(sales_data_2017,sales_data_2018,on=["Party","Product"],how="left",sort=False)
sales_edited_data
```

#### 4. right_join

```python
sales_edited_data=pd.merge(sales_data_2017,sales_data_2018,on=["Party","Product"],how="right",sort=False)
sales_edited_data
```

#### 5.left_on and_right_on 1.inner - left_on & right_on

```python
sales_edited_data=pd.merge(sales_data_2017,sales_data_2018,left_on="Party",right_on="Product",how="inner",sort=False)
sales_edited_data
```

#### 2.outer=left_on & right_on

```python
sales_edited_data=pd.merge(sales_data_2017,sales_data_2018,left_on="Party",right_on="Product",how="outer",sort=False)
sales_edited_data
```

#### 3.left=left_on & right_on

```python
sales_edited_data=pd.merge(sales_data_2017,sales_data_2018,left_on="Party",right_on="Product",how="left",sort=False)
sales_edited_data
```

#### 4.right=left_on & right_on

```python
sales_edited_data=pd.merge(sales_data_2017,sales_data_2018,left_on="Party",right_on="Product",how="right",sort=False)
sales_edited_data
```

## lambda

```python
from functools import reduce
```

```python
employee_data=pd.read_csv("employee_data.csv")
employee_data
```

```python
employee_data["current values"]=employee_data.apply("Salary").map(lambda x: x * 2).round()
employee_data
```

```python
employee_data["current values"]=employee_data.apply("Salary").filter([lambda x: x * 2]).round()
employee_data
```

```python
employee_data["current values"]= reduce(lambda result,x:result+[round(x*2)],employee_data["Salary"].fillna(0),[])
employee_data
```
